# MATH 331 - Differential Equations: notebook library

A library of Jupyter notebooks for the MATH 331 lecture series, following the course's
**Lecture Plan** (`ode/MATH331_Lecture_Plan.pdf`) and the open textbook adopted for the class:

> **J. Lebl, *Notes on Diffy Qs: Differential Equations for Engineers*** (CC BY-SA, free on
> LibreTexts), file `ode/FullText_Diff_Eq_For_Engineers.pdf`. Section numbers in the notebooks
> refer to this edition.

Every notebook follows the same structure:

1. **the reference problem**, stated precisely, with its modelling assumptions;
2. classification, and **existence and uniqueness** (Picard's theorem, Lebl Theorem 1.2.1);
3. the **analytical solution**, derived step by step and verified symbolically with SymPy;
4. worked examples **reproduced from the references**, with `assert` statements that stop the
   notebook if a number does not match;
5. **numerical solutions** compared with the exact ones;
6. **sliders** (ipywidgets) to change parameters and intervals. Each slider plot has a static
   version for readers without a running kernel;
7. interpretation, limits of the model, and exercises from the references.

## Where these topics sit in the course

The lecture plan has 30 sessions over 15 units, with a *numerical thread* (Euler's method
previewed in Week 4, computational questions in every problem set) and an *applications
thread* (population models in Weeks 4 and 10). The four notebooks above cover:

* **Unit 1** (definitions, verifying solutions, IVPs, direction fields): Sessions 3;
* **Unit 2** (separable equations, linear equations, integrating factor, existence and
  uniqueness): Sessions 4-5;
* **Unit 4** (growth and decay, logistic equation, autonomous equations, phase line): Session 7;
* the **Euler preview** of Session 8 and part of **Unit 14** (Session 29).

Results in the notebooks are checked against these books from the `ode/` folder (values that

Two small slips found in the course text while cross-checking are pointed out where they occur
(notebook 04): a rounding difference in Table 1.7.1, and "$y(2)$" for "$y(1)$" in the answer to
Exercise 1.7.9. Checking every number is part of doing mathematics.

## Getting started

```bash
cd ode
conda env create -f environment.yml      # once
conda activate ode
cd notebooks
jupyter lab                              # or open the folder in VS Code
python -m pytest                         # tests of the helper library odelib.py
```

The helper module `odelib.py` (in this folder) holds the tested building blocks: analytical
solutions, fixed-step Euler/Heun/RK4 solvers, slope fields and phase lines. Its tests reproduce
the reference values above.
